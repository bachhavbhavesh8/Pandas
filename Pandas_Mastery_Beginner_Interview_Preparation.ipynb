# Series

A Series is a one-dimensional labelled array capable of holding heterogeneous data.

## Components
- Data
- Index
- Data Type

## Facts
- Mutable values
- Immutable size
- Supports vectorized operations

```python
# Create Series from list
s1 = pd.Series([10,20,30], index=['a','b','c'])
print(s1)

# Create Series from dictionary
s2 = pd.Series({'Math':90,'Science':95,'English':88})
print(s2)
```

# DataFrame

A DataFrame is a two-dimensional labelled data structure.

## Properties
1. Heterogeneous Data
2. Size Mutable
3. Labelled Axes
4. Supports Missing Values
5. Powerful Indexing

```python
# List of lists
df1 = pd.DataFrame(
    [['John',25],['Mary',30]],
    columns=['Name','Age']
)
print(df1)

# List of dictionaries
df2 = pd.DataFrame([
    {'Name':'Alice','Age':28},
    {'Name':'Bob','Age':35}
])
print(df2)

# Dictionary of Series
df3 = pd.DataFrame({
    'A': pd.Series([1,2,3]),
    'B': pd.Series([4,5,6])
})
print(df3)
```

# Adding Columns

Methods:
- Assign scalar value
- Assign list
- Assign Series

```python
df = pd.DataFrame({
    'Name':['John','Mary','Alice'],
    'Age':[25,30,28]
})

# Scalar
df['Country'] = 'India'

# List
df['Score'] = [80,90,85]

# Series 
bonus = pd.Series([5,10,15])
df['Bonus'] = bonus

print(df)
```

# Rename Columns

rename(columns={}, inplace=False)

Parameters:
- columns
- inplace

```python
# Rename single column
df.rename(columns={'Age':'Employee_Age'}, inplace=True)

# Rename multiple columns
df.rename(columns={
    'Score':'Exam_Score',
    'Bonus':'Extra_Bonus'
}, inplace=True)

print(df)
```

# Indexing and Data Selection

## Types
- loc (label based)
- iloc (position based)

```python
employees = pd.DataFrame({
    'Name':['John','Mary','Alice','Bob'],
    'Department':['IT','HR','IT','Finance'],
    'Salary':[50000,60000,55000,70000]
}, index=['E1','E2','E3','E4'])

print(employees)
```

## loc Examples

```python
# Access row
print(employees.loc['E1'])

# Access column
print(employees.loc[:, 'Name'])

# Multiple rows and columns
print(employees.loc[['E1','E3'], ['Name','Salary']])

# Conditional indexing
print(employees.loc[employees['Salary'] > 55000])
```

## iloc Examples

```python
# Single row
print(employees.iloc[0])

# Single column
print(employees.iloc[:,0])

# Subset
print(employees.iloc[0:3, 0:2])
```

# Interview Questions

# 1. Difference between Series and DataFrame?
| Series                     | DataFrame                      |
| -------------------------- | ------------------------------ |
| 1-dimensional              | 2-dimensional                  |
| Single column of data      | Multiple columns of data       |
| Has one index              | Has row index and column names |
| Created with `pd.Series()` | Created with `pd.DataFrame()`  |


# 2. loc vs iloc?
| `loc`                        | `iloc`                     |
| ---------------------------- | -------------------------- |
| Uses labels                  | Uses integer positions     |
| Can use conditions           | Cannot directly use labels |
| End label included in slices | End position excluded      |


# 3. Why is Pandas built on NumPy?

Pandas uses NumPy arrays internally because NumPy provides:

Fast numerical computations
Efficient memory usage
Vectorized operations
Powerful indexing and slicing

# 10. When would you choose loc over iloc?

Use loc when thinking in terms of names/labels.
A label is the name you give to a row or column.

Use iloc when thinking in terms of positions (0, 1, 2, ...)
A position is the numeric location (like array index).

```python
df = pd.DataFrame({
    "name":["a","b"],
    "age":[1,2]
})
print(df)
```
