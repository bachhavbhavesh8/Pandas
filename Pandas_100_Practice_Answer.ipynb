```python
### Q3. Access elements using index labels.

import pandas as pd

s = pd.Series(
    [85, 90, 78, 92],index=["a", "b", "c", "d"]
)

# Single value
print(s["a"])

# Multiple values
print(s[["a", "b"]])

# Range of labels (inclusive)
print(s["a":"c"])
```

```python
### Q4. Perform arithmetic operations on a Series.

s = pd.Series([10, 20, 30, 40, 50])

print(s + 5)    # Addition
print(s - 5)    # Subtraction
print(s * 2)    # Multiplication
print(s / 2)    # Division
print(s // 3)   # Floor division
print(s % 3)    # Modulus
print(s ** 2)   # Power
```

```python
### Q5. Find mean, min, max of a Series.

s = pd.Series([10, 20, 30, 40, 50])

print("Mean:", s.mean())
print("Min :", s.min())
print("Max :", s.max())
```

```python
# ### Q6. Apply conditional filtering on a Series.
# >**Hint:** Use a boolean mask: `s[s > 50]`. You can also chain condition  s with `&` and `|`.

import pandas as pd

s = pd.Series([25, 60, 45, 80, 55, 30])

# Values greater than 50
print(s[s > 50])


# Values between 40 and 70
print(s[(s > 40) & (s < 70)])

# Values less than 30 OR greater than 70
print(s[(s < 30) | (s > 70)])
```

```python
# ### Q49. Replace substring in column.
# > 💡 **Hint:** `df["col"].str.replace("old", "new")`.
# Add `regex=True` for pattern-based replacement.

import pandas as pd

df = pd.DataFrame({
    "Email": ["john@gmail.com", "mary@gmail.com", "alice@yahoo.com", "bob@gmail.com"]
})

print(df)
```

```python
# # 1. Replace simple substring

# Replace "gmail.com" with "company.com"

df["Email"] = df["Email"].str.replace("gmail.com", "company.com")

print(df)
```

```python
# 2. Replace using pattern (regex)
# If you want to replace all domains, use regex:
# @.* - everything after @
# Python raw strings (r"")
df["Email"] = df["Email"].str.replace(r"@.*", "@company.com", regex=True)

print(df)
```

```python
# ### Q56. Detect outliers in a column.
# > 💡 **Hint:** Use the **IQR method**: compute Q1, Q3, IQR = Q3–Q1.
# Outliers are values below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`. 
# Alternatively use Z-score: `|z| > 3`.

import pandas as pd

df = pd.DataFrame({
    "Salary": [30000, 35000, 40000, 45000, 50000, 200000]
})

print(df)
```

```python
#  Calculate Q1, Q3, and IQR
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
```

```python
# Calculate Bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(lower_bound)
print(upper_bound)
```

```python
# Detect Outliers
outliers = df[
    (df["Salary"] < lower_bound) |
    (df["Salary"] > upper_bound)
]

print(outliers)
```

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

outliers = df[
    (df["Salary"] < Q1 - 1.5 * IQR) |
    (df["Salary"] > Q3 + 1.5 * IQR)
]

print(outliers)
```

```python
# ### Q57. Handle outliers using clipping.
# > 💡 **Hint:** `df["col"].clip(lower=lo, upper=hi)` replaces values
# outside the bounds with the boundary values.

import pandas as pd

df = pd.DataFrame({
    "Salary": [30000, 35000, 40000, 45000, 50000, 200000]
})

Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df["Salary"] = df["Salary"].clip(
    lower=lower_bound,
    upper=upper_bound
)

print(df)
```

```python
import pandas as pd

# 1. LOAD DATA

df = pd.read_csv("data.csv")   # replace with your file

# ---------------------------
# 2. INSPECT DATA
# ---------------------------
print(df.head())
print(df.shape)
print(df.info())
print(df.isnull().sum())

# ---------------------------
# 3. CLEAN DATA
# ---------------------------

# Fill missing values (numeric example)
df.fillna(0, inplace=True)

# Remove duplicates
df.drop_duplicates(inplace=True)

# ---------------------------
# 4. FEATURE ENGINEERING
# ---------------------------

# Example: create new column
df["Total"] = df["Price"] * df["Quantity"]

df["Discounted_Total"] = df["Total"] * 0.9

# ---------------------------
# 5. GROUPBY ANALYSIS
# ---------------------------

category_summary = df.groupby("Category")["Total"].sum()

print(category_summary)

# ---------------------------
# 6. PIVOT TABLE ANALYSIS
# ---------------------------

pivot = pd.pivot_table(
    df,
    values="Total",
    index="Category",
    columns="Region",
    aggfunc="sum",
    fill_value=0,
    margins=True,
    margins_name="TOTAL"
)

print(pivot)

# ---------------------------
# 7. SORTING & RANKING
# ---------------------------

df_sorted = df.sort_values(by="Total", ascending=False)

df["Rank"] = df["Total"].rank(ascending=False)

# ---------------------------
# 8. FINAL INSIGHTS
# ---------------------------

top_categories = df.groupby("Category")["Total"].sum().sort_values(ascending=False)

print(top_categories)

# ---------------------------
# 9. EXPORT RESULTS
# ---------------------------

df.to_csv("cleaned_data.csv", index=False)
pivot.to_csv("summary_report.csv")

with pd.ExcelWriter("final_report.xlsx") as writer:
    df.to_excel(writer, sheet_name="Cleaned_Data", index=False)
    pivot.to_excel(writer, sheet_name="Pivot_Summary")

print("Analysis completed and files exported successfully.")
```
